```python
df["hora_num"] = pd.to_datetime(df["hora"], format="%H:%M:%S").dt.hour
```

```python
df[["hora", "hora_num"]].head()
```

```python
def clasificar_hora(hora):

    if 0 <= hora < 6:
        return "Madrugada"

    elif 6 <= hora < 12:
        return "Mañana"

    elif 12 <= hora < 18:
        return "Tarde"

    else:
        return "Noche"
```

```python
df["franja_horaria"] = df["hora_num"].apply(clasificar_hora)
```
